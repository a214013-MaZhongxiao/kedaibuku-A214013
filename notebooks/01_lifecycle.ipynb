## Data Engineering Lifecycle – KedaiBuku.my

| Lifecycle Stage | What happens at KedaiBuku.my | Tool we will use |
|---|---|---|
| Generation | Customers buy books and leave reviews on the website. | Website, Database |
| Ingestion | Collect book sales, customer reviews and exchange rate data. | Python, APIs |
| Storage | Store the collected data for future use. | PostgreSQL |
| Transformation | Clean the data, convert prices to MYR and calculate total sales. | Python, dbt |
| Serving | Show sales reports and customer reviews in a dashboard. | SQL, Streamlit |
| Undercurrents (security, governance, orchestration) | Protect customer data, manage data quality and schedule tasks. | GitHub, Airflow |
