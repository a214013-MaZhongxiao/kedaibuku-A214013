```python
import pandas as pd
from pathlib import Path

root = Path.cwd()
if not (root / "data" / "lake" / "bronze").exists():
    root = root.parent

bronze = root / "data" / "lake" / "bronze"

books = pd.read_csv(bronze / "books.csv")
customers = pd.read_csv(bronze / "customers.csv")
orders = pd.read_csv(bronze / "orders.csv")
reviews = pd.read_csv(bronze / "reviews.csv")

for name, df in [
    ("books", books),
    ("customers", customers),
    ("orders", orders),
    ("reviews", reviews)
]:
    print(f"{name}: {df.shape}")
    display(df.head())
```

### Step 4: Questions and Answers

**Q1. What does one row mean in each file?**

- books.csv: One row represents one book.
- customers.csv: One row represents one customer record.
- orders.csv: One row represents one order.
- reviews.csv: One row represents one customer review.

**Q2. Which columns link the files together?**

We use `customer_id` to link customers with orders and reviews. We use `book_id` to link books with orders and reviews.

**Q3. Which columns hold personal data?**

In customers.csv, the personal data columns are `full_name`, `email`, `phone`, and `mykad`. We should protect this information.

**Q4. Which source would you collect once, and which every day?**

I would collect the book catalogue once at the beginning and update it when needed. I would collect exchange rates every day because they can change. Customer, order and review data should also be collected regularly because new records can be added.
