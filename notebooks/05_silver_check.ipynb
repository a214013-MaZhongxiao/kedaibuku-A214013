```python
import pandas as pd

# Load original customer data
bronze = pd.read_csv("../data/lake/bronze/customers.csv")

# Load cleaned customer data
silver = pd.read_parquet("../data/lake/silver/customers.parquet")

# Show original data
print("Bronze Customers - First 5 Rows:")
display(bronze.head())

# Show cleaned data
print("Silver Customers - First 5 Rows:")
display(silver.head())

# Compare row counts
print("Bronze rows:", len(bronze))
print("Silver rows:", len(silver))



# Load the cleaned books data
books = pd.read_parquet("../data/lake/silver/books.parquet")

# Show the first 5 rows with selected columns
display(books[["title", "price_gbp", "price_myr",
               "fx_rate_date", "rating"]].head())
```

The Bronze dataset has 515 rows, while the Silver dataset has 500 rows because duplicate records were removed. Personal information such as names, phone numbers and MyKad numbers was also removed to protect customer privacy.

The Silver books dataset shows the cleaned book information. The prices are converted from GBP to MYR using the exchange rate. The ratings are also changed from words into numbers, making the data easier to analyse.
