### Q1. Quarantine versus delete
We keep bad orders in a separate file because we may need to check and fix them later. Deleting them would lose useful information. In a real company, a data engineer or data quality team would check the file.

### Q2. Hash versus mask
Hashing allows us to identify the same email without showing the real email. Masking hides part of the email for display. Without a salt, attackers could guess common email addresses and compare their hashes.

### Q3. Data minimisation
This follows the PDPA 2010 General Principle of processing personal data only for a lawful and necessary purpose. Age groups are safer because they give less personal detail than exact birth years, so it is harder to identify a person.
